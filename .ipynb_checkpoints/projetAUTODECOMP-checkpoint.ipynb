```python
from scipy.io.wavfile import read
import numpy as np
from numpy.fft import fft, fftshift, ifft, fftfreq
import matplotlib.pyplot as plt
import scipy.signal as signal
import scipy
import math
import librosa
```

```python
rate, x = read("piano_mix.wav")
print(rate)
t = np.linspace(0, x.size, x.size)
plt.plot(t,x);
```

## Etape 1 : faire la TF (facile) :)

```python
X = np.fft.fft(x)

freqs = fftshift(fftfreq(X.size, 1/rate))  #axedesX

X=np.fft.fftshift(X)
plt.xlim(-5000,5000)
plt.plot(freqs, np.abs(X));
```

```python
def divergence(V,W,H, beta = 2):
    
    """
    beta = 2 : Euclidean cost function
    beta = 1 : Kullback-Leibler cost function
    beta = 0 : Itakura-Saito cost function
    """ 
    
    if beta == 0 : return np.sum( V/(W@H) - math.log10(V/(W@H)) -1 )     bêta = 2 : fonction de coût euclidienne
    
    if beta == 1 : return np.sum( V*math.log10(V/(W@H)) + (W@H - V))    bêta = 1 : fonction de coût de Kullback-Leibler
    
    if beta == 2 : return 1/2*np.linalg.norm(W@H-V) beta = 0 : fonction de coût dItakura-Saito
```
