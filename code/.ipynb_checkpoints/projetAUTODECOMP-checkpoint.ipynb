## Etape 1 : faire la TF (facile) :)

```python
X = np.fft.fft(x)

freqs = fftshift(fftfreq(X.size, 1/rate))  #axedesX
print(freqs)

X=np.fft.fftshift(X)
plt.xlim(-5000,5000)
plt.plot(freqs, np.abs(X));
```

```python

def plot_spectrogramme(frequencies, times, spectrogram, title="Spectrogram"):

    magnitude = np.abs(spectrogram)
    plt.figure(figsize=(10, 6))
    plt.pcolormesh(times, frequencies, 20 * np.log10(magnitude+1), shading='gouraud') #20 log de magnitude = pour mettre l'amplitude en dB
    plt.colorbar(label="Amplitude (dB)")
    plt.title(title)
    plt.xlabel("Temps (s)")
    plt.ylabel("Fréquence (Hz)")
    plt.show()

    
def creer_fenetre(taille_fenetre, debut_fenetre, zeros):
    hamming = np.hamming(taille_fenetre)
    for i in range(zeros.size):
        if i > debut_fenetre and i < debut_fenetre +taille_fenetre:
            zeros[i]=hamming[i-debut_fenetre]
    return zeros


def calculer_spectrogramme(x, taille_fenetre,chevauchement):
    
    zeros = np.zeros(x.size)
    debut_fenetre = 0
    nb_fenetres = x.size // (taille_fenetre-chevauchement)
    
    spectrogramme = np.zeros((x.size, nb_fenetres ))  #remplacer 
    
    for i in tqdm(range(nb_fenetres)):
        
        fenetre = creer_fenetre(taille_fenetre, debut_fenetre, zeros)

        colonne = x * fenetre
        fft = np.fft.fftshift(np.fft.fft(colonne))
        spectrogramme[:,i] = fenetre
        
        debut_fenetre = debut_fenetre + taille_fenetre-chevauchement
        
    return spectrogramme

#marche : 
largeur_colonne = 2000
taille_fenetre = 6000
points_en_commun = 4000
spectrogramme = calculer_spectrogramme(x, taille_fenetre,points_en_commun)

#f, t, spectrogramme = scipy.signal.stft(x, fs=rate, nperseg = taille_fenetre, noverlap = points_en_commun)
#plot_spectrogramme(f, t, spectrogramme)


print(freqs.shape, x.size //taille_fenetre, np.shape(spectrogramme))

f=np.array(freqs)
nb_fenetres=x.size//(taille_fenetre-points_en_commun)
duree=x.size/rate
t=np.linspace(1,duree,nb_fenetres)
s=np.array(spectrogramme)
#print(freqs.shape,t.shape,spectrogramme.shape)
print(x.size/rate)
plot_spectrogramme(np.abs(freqs), t  , np.abs(spectrogramme))
```
