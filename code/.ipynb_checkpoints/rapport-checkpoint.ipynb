# Première partie: Importation du signal et spectrogramme <a class="anchor" id="P1"></a>

#### Fichier son  <a class="anchor" id="Fson"></a>

Pour analyser un fichier son avec python, nous utilisons d’abord la fonction [`read`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.io.wavfile.read.html) du module *scipy*. Cette fonction retourne deux paramètres : la fréquence d'échantillonnage, et les données sous forme de matrice de taille $1 * nbdonnées$. Le nombre de données correspond à la durée de l’audio multipliée par la fréquence d'échantillonnage, et chaque donnée correspond à la fréquence enregistrée sur une période de temps très petite (1/fe secondes). Nous pouvons déjà visualiser le signal temporel en traçant les données avec la fonction [`plot`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html) du module pyplot de matplotlib. 




```python
import scipy 
import matplotlib.pyplot as plt
import numpy as np
from scipy.io.wavfile import read


def chargement_audio(nom_fichier):
    rate, son = read("Fichiers_sonores/" + nom_fichier )
    return son, rate

son, rate = chargement_audio("piano_mix.wav" )

def afficher_signal(x,y, axe_x, axe_y, titre):
    plt.xlabel(axe_x)
    plt.ylabel(axe_y)
    plt.title(titre)
    plt.plot(y,x)


t = np.linspace(0, son.size, son.size)
afficher_signal(son, t, "Temps(s)", "Fréquence(Hz)", "Signal temporel" )
```

#### Transformée de Fourier  <a class="anchor" id="TF"></a>
Cet affichage nous permet de visualiser le signal dans le temps. 

Cependant, cela ne nous avance pas du tout pour le décomposer. C’est pourquoi nous utilisons cet outil que nous connaissons déjà bien : la Transformée de Fourier. Nous utilisons les fonctions [`fft`](https://numpy.org/doc/stable/reference/generated/numpy.fft.fft.html) et [`fftshift`](https://numpy.org/doc/stable/reference/generated/numpy.fft.fftshift.html) du module fft de numpy. Ainsi, nous obtenons une représentation du signal, non plus en fonction du temps mais des fréquences. Nous l’affichons comme précédemment. 



#### Spectrogramme
La deuxième étape est cruciale. À partir du signal temporel, nous allons réaliser un spectrogramme. 

**Mais qu'est-ce qu'un spectrogramme ?**

Dans notre cas, c'est une matrice contenant 3 informations. Pour chaque temps et chaque fréquence, nous avons la "quantité d'occurence". Par exemple, si entre 2s et 3s sur notre signal on a 3 fois une fréquence de 2000Hz, nous aurons dans la matrice à l'intersection de ces deux valeurs une case avec un 3.

**Comment ça marche ?**

Pour réaliser le spectrogramme, nous allons modéliser une fenêtre qui mesurera seulement une partie de notre signal, et nous allons la multiplier par notre signal afin d'en garder seulement une partie qui correspondra à un intervalle. Dans l'exemple précédent, cet intervalle durera 1 seconde. Ensuite, nous allons appliquer au signal multiplié une ['rfft'](https://numpy.org/doc/stable/reference/generated/numpy.fft.rfft.html) du module numpy. Cette opération est la *Real Fast Fourier Transform*, c'est-à-dire la Transformée de Fourier Discrète sur des valeurs réelles (comme on a un signal audio réel, on peut l'appliquer).
Enfin, nous insérons ce résultat dans la colonne du spectrogramme correspondante à la fenêtre.
Et ensuite, nous faisons "glisser" notre fenêtre, c'est à dire qu'on lui ajoute un pas, afin d'avoir, à la fin analysé tout le signal.



Nous avons dans un premier temps utilisé le code ['spectrogram'](https://docs.scipy.org/doc/scipy/reference/generated/scipy.signal.spectrogram.html) de la bibliothèque scipy.signal. Puis nous l'avons codé à la main.
Voici les résultats de ces deux méthodes.

```python
#Spectrogramme 
def plot_spectrogramme(frequencies, times, spectrogram, title="Spectrogram"):

    magnitude = np.abs(spectrogram)
    plt.figure(figsize=(10, 6))
    plt.pcolormesh(times, frequencies, 10 * np.log10(magnitude+1), shading='gouraud') #20 log de magnitude = pour mettre l'amplitude en dB
    plt.colorbar(label="Amplitude (dB)")
    plt.title(title)
    plt.xlabel("Temps (s)")
    plt.ylabel("Fréquence (Hz)")
    plt.show()

def calculer_spectrogramme(x, rate, taille_fenetre, saut, fenetre):

    #création d'une fenêtre de hanning
    fenetre = fenetre(taille_fenetre)

    # Nombre de fenêtres
    nb_fenetres = (x.size - taille_fenetre) // saut + 1

    # Initialisation du spectrogramme
    spectrogramme = np.zeros((taille_fenetre // 2 + 1, nb_fenetres))
    times = np.arange(nb_fenetres) * saut / rate
    frequencies = np.fft.rfftfreq(taille_fenetre, d=1/rate)

    # Calcul des TF pour chaque segment
    for i in range(nb_fenetres):
        debut = i * saut
        fin = debut + taille_fenetre
        # on prend une partie de x
        echantillon = x[debut:fin] 
        # on applique un filtre en forme de fenetre de hanning
        echantillon_filtre = echantillon * fenetre
        # on applique une Real Fast Fourier Transform (car on a un signal audio, donc réel, donc on a pas besoin des complexes)
        transformee = np.fft.rfft(echantillon_filtre)
        # on l'ajoute dans une colonne du spectrogramme :)
        spectrogramme[:, i] = np.abs(transformee)

    return frequencies, times, spectrogramme


# Paramètres communs
taille_fenetre = 1024
saut = 512

# Méthode 1 : scipy.signal.spectrogram
f1, t1, Sxx = scipy.signal.spectrogram(son, fs=rate, window="hann", nperseg=taille_fenetre, noverlap=saut)

# Méthode 2 : le code à la main 
f3, t3, Sxx_a_la_main = calculer_spectrogramme(son, rate, taille_fenetre, saut, np.hanning)

plt.figure(figsize=(12, 5))

plt.subplot(121)
plt.pcolormesh(t1, f1, 10 * np.log10(Sxx), shading='gouraud')
plt.title("Spectrogramme avec \n scipy.signal.stft")
plt.ylabel("Fréquence (Hz)")
plt.xlabel("Temps (s)")
plt.colorbar(label="Amplitude (dB)")


plt.subplot(122)
plt.pcolormesh(t3, f3, 10 * np.log10(Sxx_a_la_main), shading='gouraud')
plt.title("Spectrogramme avec fenêtre \n glissante, à la main")
plt.ylabel("Fréquence (Hz)")
plt.xlabel("Temps (s)")
plt.colorbar(label="Amplitude (dB)")

plt.tight_layout()
plt.show()
```

# Partie 2 : Factorisation en Matrices Non-Négatives (NMF)  <a class="anchor" id="P2"></a>

Dans le cadre de la séparation de sources sonores, la Factorisation en Matrices Non-Négatives (NMF) joue un rôle crucial. Cette méthode mathématique permet de décomposer une matrice contenant des données complexes en produit de deux matrices de rang inférieur **V=WH**, tout en conservant la contrainte de non-négativité. Appliquée au spectrogramme d'un signal audio, la NMF facilite l'extraction des sources individuelles à partir d'un enregistrement monophonique.

**Pourquoi la NMF est utile pour notre séparation des sources ?**


Dans notre cas, la nature des matrices trouvées à l'issue de la NMF permet de faciliter la manipulation des sources individuelles. En effet, la matrice W représente notre **dictionnaire** qui contient pour chaque colonne (et donc pour chaque source) son **"identité fréquentielle"**. 

La matrice H quant à elle représente notre **matrice d'activation** qui contient pour chaque ligne (et donc pour chaque source) son **activation temporelle**.

Ainsi, pour obtenir le signal uniquement pour notre ième source, il suffit de faire le produit extérieur de la ième colonne de W avec la ième ligne de H pour construire un spectrogramme propre à notre source puis retourner en domaine temporel pour construire le signal de notre source.


#### Principe de la NMF  <a class="anchor" id="TF"></a>


La décomposition NMF permet de décomposer une matrice V en produit de deux sous matrices : V=W.H (approximation)
On notera par la suite la taille des matrice :  &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;  **V : m x n**    &nbsp;&nbsp;   **W : m x k**    &nbsp;&nbsp;      **H: k x n**

Ici, on voit bien que que l'on ne connait pas k (nombre de composantes latentes) qui est pourtant essentiel à notre décomposition, on l'introduit donc en parametre **noté R** qu'on cherchera à optimiser par la suite.


Afin de trouver la meilleure approximation (ayant le paramètre R fixé) il faudrait pouvoir évaluer la similarité entre V et WH. On utilise la β-divergence qui est une mesure de dissimilarité que l'on cherchera à minimiser.



La **β-divergence** entre \( V \) et \( WH \) est donnée par :

$$
D_\beta(V, WH) = \frac{1}{\beta(\beta - 1)} \sum_{i,j} \left( V_{ij}^\beta + (\beta - 1) (WH)_{ij}^\beta - \beta V_{ij} (WH)_{ij}^{\beta-1} \right)
$$

Les cas particuliers sont :

- Pour β = 2   (divergence quadratique) :
  
$$
D_2(V, WH) = \sum_{i,j} (V_{ij} - (WH)_{ij})^2
$$

- Pour β = 1   (divergence de Kullback-Leibler) :
  
$$
D_1(V, WH) = \sum_{i,j} \left( V_{ij} \log \frac{V_{ij}}{(WH)_{ij}} - V_{ij} + (WH)_{ij} \right)
$$

- Pour β = 0   (divergence de Itakura-Saito) :
  
$$
D_0(V, WH) = \sum_{i,j} \left( \frac{V_{ij}}{(WH)_{ij}} - \log \frac{V_{ij}}{(WH)_{ij}} - 1 \right)
$$



**Comment la minimiser ?**

Cette distance dépend de 2 variables (W et H), afin de la minimiser on peut utiliser plusieurs methodes, comme la méthode de descente du gradient (au sens des moindres carrés) ou de mise à jour alternée multiplicative. Nous avons choisi la méthode de mise à jour alternée multiplicative dont le principe est le suivant:





- **1. Initialisation :**
On initialise les matrices $W^{(0)}$ et $H^{(0)}$ de manière aléatoire.

- **2. Mise à jour de  H ( W fixé ) :**

$$
H^{(l+1)} = H^{(l)} \times \frac{W^T \cdot \left( \left( W \cdot H^{(l)} \right)^{\beta-2} \cdot V \right)}{W^T \cdot \left( W \cdot H^{(l)} \right)^{\beta-1} + \epsilon}
$$

- **3. Mise à jour de W ( H fixé ) :**

$$
W^{(l+1)} = W^{(l)} \times \frac{\left( \left( W^{(l)} \cdot H^{(l)} \right)^{\beta-2} \cdot V \right) \cdot H^{(l)^T}}{\left( W^{(l)} \cdot H^{(l)} \right)^{\beta-1} \cdot H^{(l)^T} + \epsilon}
$$

- **4. Critère d'arrêt :**

L'algorithme répète les étapes 2 et 3 jusqu'à ce que la fonction de coût converge, c'est-à-dire que la différence entre deux itérations successives soit inférieure à un seuil \( \epsilon \) :

$$
|D_\beta(V, W^{(l+1)} H^{(l+1)}) - D_\beta(V, W^{(l)} H^{(l)})| < \epsilon
$$

- **Remarques :**
- β détermine la divergence utilisée, par exemple, pour $ \beta = 2 $ on a la divergence Euclidienne (erreur quadratique).
- En prenant $W^{(0)}$ et $H^{(0)}$ non négatives, on est sur de n'avoir que des matrices non négatives car les mises à jour de W et H n'utilisent que des multiplications. C'est pour cela que l'on a opté pour cette méthode.


#### Application dans le Projet

1-**Prétraitement du Signal**
Avant d'appliquer la NMF, un spectrogramme est calculé à partir du signal audio brut en utilisant une Transformée de Fourier à Court Terme (STFT). Ce spectrogramme, représenté sous forme de matrice, sert d'entrée pour la NMF.

2-**Décomposition NMF**
La méthode vise à décomposer le spectrogramme en matrices  et  qui modélisent respectivement les caractéristiques spectrales et temporelles des sources. Cela permet de distinguer les différentes composantes sonores présentes dans le mélange.

3-**Reconstruction des Sources**
Une fois  et  obtenues, les spectres individuels des sources peuvent être reconstruits en multipliant chaque colonne de  par la ligne correspondante de . Le retour au domaine temporel est effectué via une Transformée de Fourier inverse.

#### Avantages de la NMF

-**Simplicité d'Interprétation** : La contrainte de non-négativité rend les matrices  et  facilement interprétables dans des contextes physiques (par exemple, énergie ou puissance).

-**Robustesse** : La NMF est bien adaptée à la séparation de sources dans des mélanges polyphoniques complexes.

-**Flexibilité** : Elle peut être combinée avec d'autres méthodes, comme les régularisations ou les contraintes supplémentaires, pour améliorer la qualité de la décomposition.

### Implémentation

Fonction de la beta divergence

```python
def divergence(V,W,H, beta ):
    
    """
    beta = 2 : Norme de Frobenius (erreur quadratique)
    beta = 1 : Divergence de Kullback-Leibler (KL) 
    beta = 0 : Divergence Itakura-Saito (IS)
    """ 
    
    if beta == 0 : return np.sum( V/(W@H) - np.log10(V/(W@H)) -1 )
    
    if beta == 1 : return np.sum( V*np.log10(V/(W@H)) + (W@H - V))
    
    if beta == 2 : return 1/2*np.linalg.norm(W@H-V)
```

Initialisation de la NMF

```python
def init_NMF(son_stft_Magnitude):
    epsilon = 1e-10 # erreur 
    V = son_stft_Magnitude + epsilon
    return V
```

Fonction NMF 

```python
def NMF(V, S, beta , afficher_spectro,rate,SAUT, epsilon = 0.05, MAXITER = 5000, displayEveryNiter = None): 
    
    """
    Entrée : 
    --------
    
        V         : Signal de mélange : |TFST| (par exemple, un spectrogramme)
        S         : Le nombre de sources à extraire
        beta      : Divergence beta utilisée, par défaut = 2 (Euclidienne)
        afficher_specto : Permet de choisir si les spectrogrammes doivent être affichés ou non
        epsilon   : Critère d'arrêt (valeur seuil), par défaut = 0.05
        MAXITER   : Nombre maximal d'itérations, par défaut = 5000
        displayEveryNiter : Indique si les affichages intermédiaires doivent être faits uniquement à la dernière itération
    
    Sorties :
    ---------
      
        W : Matrice de dictionnaire [KxS], W>=0
        H : Matrice d'activation [SxN], H>=0
        fonction_cout : Fonction de coût optimisée sur les itérations
       
    Algorithme : 
    -----------
   
    1) Initialisation aléatoire des matrices W et H
    2) Mise à jour multiplicative des matrices W et H 
    3) Répéter l'étape (2) jusqu'à la convergence ou après MAXITER itérations 
    
    """
    compteur  = 0
    fonction_cout = []
    beta_divergence = 1
    
    K, N = np.shape(V)
    
    # On initialise de manière aléatoire W0 et H0 (l=0)
    W = np.abs(np.random.normal(loc=0, scale = 2.5, size=(K,S)))    
    H = np.abs(np.random.normal(loc=0, scale = 2.5, size=(S,N)))
    
    # Affichage de la première itération
    #plot_NMF_iter(W,H,beta,compteur)


    while beta_divergence >= epsilon and compteur <= MAXITER:
        
        # Mise à jour de W et H
        H *= (W.T@(((W@H)**(beta-2))*V))/(W.T@((W@H)**(beta-1)) + 10e-10) #on rajoute un epsilon pour ne jamais diviser par 0
        W *= (((W@H)**(beta-2)*V)@H.T)/((W@H)**(beta-1)@H.T + 10e-10)
        
        W = np.maximum(W, 1e-9) #on prend le max entre W et 1e-9 pour être sûr de ne jamais prendre 0 0
        H = np.maximum(H, 1e-9)

        # Calcul de la fonction de coût
        beta_divergence =  divergence(V,W,H, beta)
        fonction_cout.append( beta_divergence )
        
        if compteur%displayEveryNiter == 0 and afficher_spectro : 
            plot_NMF_iter(W,H,beta, rate=rate , SAUT=SAUT,iteration = compteur)

        compteur +=1
    
    if compteur -1 == MAXITER : print(f"On est arrivés à {MAXITER} itérations.")
    else : print(f" La fonction converge au bout de {counter-1} iterations.")
        
    return W,H, fonction_cout
```

Filtrage des sources

```python
def filtrage_sources(W, H,V,S,afficher_spect,rate,SAUT):
    spectrogrammes_filtre = []
    if afficher_spect:
        f, axs = plt.subplots(nrows=1, ncols=S,figsize=(20,5))
    for i in range(S):
        
        # Filtrage de chaque source
        WsHs = W[:,[i]]@H[[i],:]
        spectrogramme_filtre = W[:,[i]]@H[[i],:] /(W@H *1e-9) * V 
        
        # Calculer le spectrogramme filtré
        D = librosa.amplitude_to_db(spectrogramme_filtre, ref = np.max)
        
        # Affichage du spectrogramme filtré
        if afficher_spect:
            axs[i].set_title(f"Frequency Mask of Audio Source s = {i+1}") 
            librosa.display.specshow(D,y_axis = 'hz', sr=rate,hop_length=SAUT,x_axis ='time',cmap= matplotlib.cm.jet, ax = axs[i])
        spectrogrammes_filtre.append(spectrogramme_filtre)
    return spectrogrammes_filtre
```

Reconstruction du signal

```python
def reconstruction_signal(son_stft,spectrogrammes_filtre,S,FENETRE,SAUT):
    son_stft_angle = np.angle(son_stft)
    sons_reconstruits = []
    for i in range(S):
        fourier_inverse = spectrogrammes_filtre[i] * np.exp(1j*son_stft_angle)
        source_reconstruite   = librosa.istft(fourier_inverse, n_fft = FENETRE, hop_length = SAUT)
        sons_reconstruits.append(source_reconstruite)
    return sons_reconstruits
```

Voici les codes liés à l'affichage

```python
def plot_NMF_iter(W, H,beta, rate , SAUT,iteration = None ):
    
    f = plt.figure(figsize=(4,4))
    f.suptitle(f"NMF Iteration {iteration}, for beta = {beta}", fontsize=8,)
    
    # definitions for the axes
    V_plot = plt.axes([0.35, 0.1, 1, 0.6])
    H_plot = plt.axes([0.35, 0.75, 1, 0.15])
    W_plot = plt.axes([0.1, 0.1, 0.2, 0.6])

    D = librosa.amplitude_to_db(W@H, ref = np.max)

    librosa.display.specshow(W,y_axis = 'hz', sr=rate, hop_length=SAUT,x_axis ='time',cmap= matplotlib.cm.jet, ax=W_plot)
    librosa.display.specshow(H,y_axis = 'hz', sr=rate, hop_length=SAUT,x_axis ='time',cmap= matplotlib.cm.jet, ax=H_plot)
    librosa.display.specshow(D,y_axis = 'hz', sr=rate, hop_length=SAUT,x_axis ='time',cmap= matplotlib.cm.jet, ax=V_plot)

    W_plot.set_title('Dictionnary W', fontsize=10)
    H_plot.set_title('Temporal activations H', fontsize=10)

    W_plot.axes.get_xaxis().set_visible(False)
    H_plot.axes.get_xaxis().set_visible(False)
    V_plot.axes.get_yaxis().set_visible(False)
```

### Protocole complet de séparation de sources audio : 
Dans cette fonction nous allons réaliser le protocole complet afin d'obtenir les signaux séparés. Nous avons également ajouté des options afin de choisir si on affiche les spectrogrammes et les différents graphiques.



```python
from scipy.io.wavfile import read
import numpy as np
from numpy.fft import fft, fftshift, ifft, fftfreq
import matplotlib.pyplot as plt
import matplotlib
import scipy.signal as signal
import librosa
import librosa.display
import IPython.display as ipd
from sklearn import decomposition
FENETRE = 512
SAUT = 256

def protocole_complet(fichier,S,beta, afficher_spect,afficher_son,afficher_spect_son,afficher_cout) :
    #Chargement plus affichage audio
    son, rate = chargement_audio(fichier)
    t = np.linspace(0, son.size, son.size)
    affichage_signal(son, rate)
    
    #spectrogramme avec librosa
    FENETRE = 512
    SAUT = 256
    son_stft = librosa.stft(son, n_fft = FENETRE, hop_length = SAUT)
    son_stft_Magnitude, spec  = creation_spectrogramme(son_stft,afficher_specto=afficher_spect,rate=rate,SAUT=SAUT)

    #NMF
    V = init_NMF(son_stft_Magnitude)
    W, H, fonction_cout = NMF(V,S,beta = beta,afficher_specto= afficher_spect, rate=rate , SAUT= SAUT, epsilon = 0.05, MAXITER = 10000, displayEveryNiter = 5000)   
    if afficher_cout :
        affichage_cout(fonction_cout, beta)
        
    #Spectrogramme des sons filtrés et reconstruction des signaux
    spectrogrammes = filtrage_sources(W, H,V,S,afficher_spect, rate=rate , SAUT=SAUT)
    sons_separes = reconstruction_signal(son_stft,spectrogrammes,S,FENETRE=FENETRE , SAUT=SAUT)

    #Affichage et écoute des signaux séparés
    if afficher_spect_son:
        affichage_signaux(S,sons_separes,rate=rate)
    if afficher_son : 
        ecoute_signaux_separes(S,sons_separes,rate=rate)
```

# Troisième partie: Axes d'amélioration <a class="anchor" id="P1"></a>

## Initialisation de W et H
Dans notre algorithme, ces deux matrices sont initialisées de manière aléatoire. Cela peut nous empêcher d'obtenir la meilleure solution. En effet, il est possible que l'algorithme s'arrête lorsqu'il a trouvé un minimum local. Pour contrer cela, deux solutions sont possibles:

-On lance le programme plusieurs fois avec une initialisation aléatoire, on compare les solutions obtenues et on garde la meilleure.

-On initialise W et H grâce à la Singular Value Decomposition (SVD), afin d'obtenir un point de départ proche de la solution. La SVD consiste à décomposer une matrice en vecteurs et valeurs singulières, permettant d'obtenir une approximation du rang réduit de la matrice. En eliminant les petites valeurs singulières, on peut éliminer le bruit dans les données et obtenir une meilleure décomposition finale.

## Alternating Least Square (ALS)
Lors de nos recherches sur les méthodes de décomposition, nous avons trouvé l'ALS.
Cette méthode consiste à minimiser une fonction de coût comme pour la NMF. 

La différence se trouve dans la mise à jour de W et H. On considère W fixe et on met à jour H, puis on considère H fixe et on met à jour W. Lors de cette mise à jour, W et H peuvent prendre des valeurs négatives, il faut donc mettre ces valeurs à 0 une fois la mise à jour effectuée. 

Cela peut donc causer une perte d'interpétrabilité ou de divergence puisque temporairement W et H ont des valeurs négatives.

Nous n'avons pas implémenté cette méthode puisqu'elle est moins adaptée pour les signals sonores que la NMF. Cependant, elle a l'avantage de converger plus rapidement que la NMF et être moins sensible au bruit.
